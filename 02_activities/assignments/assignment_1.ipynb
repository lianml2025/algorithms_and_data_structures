
-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
# Your answer here

def is_valid_brackets(s: str) -> bool:
    # Map closing brackets to their corresponding opening brackets
    bracket_pairs = {')': '(', '}': '{', ']': '['}
    
    # Stack to track opening brackets
    stack = []
    
    for char in s:
        if char in bracket_pairs.values():
            # Character is an opening bracket - push to stack
            stack.append(char)
        elif char in bracket_pairs:
            # Character is a closing bracket
            if not stack or stack[-1] != bracket_pairs[char]:
                return False
            stack.pop()
    
    # Valid if stack is empty (all brackets matched)
    return not stack

# Test cases
if __name__ == "__main__":
    test_cases = [
        "([]{})",    # True - proper nesting
        "([)]",      # False - wrong order  
        "()[]{}",    # True - separate valid pairs
        "[{]}",      # False - mismatched brackets
        "",          # True - empty string
        "((()))",    # True - nested same type
        "({[}])",    # False - incorrect nesting
    ]
    
    for test in test_cases:
        result = is_valid_brackets(test)
        print(f"Input: '{test}' -> Output: {result}")
````


-   Explain why your solution works


````python
# Your answer here
Using Stack to keep track of the opening brackets and ensuring the order are correct by matching with closing brackets. The mapping bracket provides the dictionary to match the closing brackets with the correct corresponding opening bracket.The steps of the logic: 1) If it's opening bracket, push them onto the stack. 2) check if the stack is empty : if yes then return false. else if top of stack matches the expeted opening bracket. If not, return false. If it matches, pop the opening bracket from the stack.3) at the end, if stack is empty, return true. else return false.
````

````python
# Your answer here
In the stack solution, we only process each character in the string once. This gives us a time complexity of O(n). 
The space complexity is also O(n) in the worst case, when all the characters in the string are openning brackets and are pushed onto the stack.
````
